### XOR Operation in an Array

```python
class Solution:
    def xorOperation(self, n: int, start: int) -> int:
        nums = [start + 2*i for i in range(n)]
        
        result = 0
        for i in nums:
          result ^= i
          
        return result
```

```python
s = Solution()

print(s.xorOperation(6,2))
```
