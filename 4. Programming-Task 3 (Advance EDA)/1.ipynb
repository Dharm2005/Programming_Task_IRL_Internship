### Maximum Product of Two Elements in an Array

```python
class Solution:
    def maxProduct(self, nums: list[int]) -> int:
        
        maximum =  max(nums)
        nums.remove(maximum)
        
        second_max = max(nums)
        
        return (maximum-1) * (second_max-1)
```

```python
s = Solution()

print(s.maxProduct([7,3,4,2]))
```
