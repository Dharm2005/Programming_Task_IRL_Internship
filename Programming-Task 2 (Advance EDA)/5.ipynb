### Subtract the Product and Sum of Digits of an Integer

```python
class Solution:
    def subtractProductAndSum(self, n: int) -> int:
        product = 1
        sum = 0
        while n > 0:
            rem = n % 10
            product *= rem
            sum += rem
            n = n // 10
            
        return product-sum
```

```python
s = Solution()

print(s.subtractProductAndSum(6548))
```
