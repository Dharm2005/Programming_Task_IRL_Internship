### Count Number of Teams

```python
class Solution:
    def numTeams(self, rating):
        n = len(rating)
        count = 0

        for j in range(n):
            l_sml = l_lrg = 0
            r_sml = r_lrg = 0

            for i in range(j):
                if rating[i] < rating[j]:
                    l_sml += 1
                else:
                    l_lrg += 1

            for k in range(j+1, n):
                if rating[k] > rating[j]:
                    r_lrg += 1
                else:
                    r_sml += 1

            count += l_sml * r_lrg
            count += l_lrg * r_sml

        return count
```

```python
s = Solution()

print(s.numTeams([1,2,5,3,6]))
```
